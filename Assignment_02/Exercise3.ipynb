# Exercise 3: Word Co-occurrence Matrix

```python
import numpy as np
```

```python
# Given sentence
sentence = "the cat sat on the mat"
```

```python
# Step 1: Split sentence into words
words = sentence.split()
print(words)
```

```python
# Step 2: Build vocabulary
vocabulary = list(dict.fromkeys(words))
print(vocabulary)
```

```python
# Step 3: Create word-to-index mapping
word_to_index = {
    word: i for i, word in enumerate(vocabulary)
}
print(word_to_index)    
```

```python
# Step 4: Create an empty co-occurrence matrix
matrix = np.zeros(
    (len(vocabulary), len(vocabulary)),
    dtype=int
)

matrix
```

```python
# Step 5: Define window size
window_size = 1
```

```python
# Step 6: Count neighboring words
for i, word in enumerate(words):

    start = max(0, i - window_size)
    end = min(len(words), i + window_size + 1)

    for j in range(start, end):

        # Don't compare a word with itself
        if i != j:

            word1 = words[i]
            word2 = words[j]

            row = word_to_index[word1]
            col = word_to_index[word2]

            matrix[row][col] += 1
```

```python
# Step 7: Display the matrix
print("Co-occurrence Matrix")

print("      ", end="")

for word in vocabulary:
    print(f"{word:>5}", end="")

print()

for i, word in enumerate(vocabulary):
    print(f"{word:>5}", end="")

    for j in range(len(vocabulary)):
        print(f"{matrix[i][j]:>5}", end="")

    print()
```
