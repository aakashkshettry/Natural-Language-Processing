# Exercise 6: Simple Keyboard Prediction

```python
# Training corpus
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP"
]
```

```python
# Dictionary to store next-word frequencies
next_words = {}
```

```python
# Build bigram counts
for sentence in corpus:

    # Split sentence into words
    words = sentence.split()

    # Create bigrams
    for i in range(len(words) - 1):

        current_word = words[i]
        next_word = words[i + 1]

        # Create dictionary for current word if it doesn't exist
        if current_word not in next_words:
            next_words[current_word] = {}

        # Count the next word
        if next_word in next_words[current_word]:
            next_words[current_word][next_word] += 1
        else:
            next_words[current_word][next_word] = 1
```

```python
# User input
user_word = "Ilike"
```

```python
# Check whether the word exists
if user_word in next_words:

    # Get possible next words
    predictions = next_words[user_word]

    # Sort by frequency
    sorted_predictions = sorted(
        predictions.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # Display top three predictions
    print("\nSuggestions:")

    for word, count in sorted_predictions[:3]:
        print(word)

else:
    print("No prediction available.")
```
