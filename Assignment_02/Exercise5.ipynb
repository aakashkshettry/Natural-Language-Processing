# Exercise 5: Implement Laplace Smoothing

```python
# Given word counts
word_counts = {
    "cat": 4,
    "dog": 3,
    "bird": 1
}
```

```python
# Step 1: Calculate total number of words
total_words = sum(word_counts.values())
```

```python
# Step 2: Calculate vocabulary size
vocabulary_size = len(word_counts)
```

```python
# Step 3: Apply Laplace smoothing
smoothed_probabilities = {}
for word, count in word_counts.items():
    probability = (count + 1) / (total_words + vocabulary_size)
    smoothed_probabilities[word] = probability
```

```python
# Step 4: Print results
print("Laplace Smoothed Probabilities:")

for word, probability in smoothed_probabilities.items():
    print(f"{word} : {probability:.4f}")
```
