# Exercise 2: Vocabulary Builder

```python
#Define the corpus
corpus = [
    "I like NLP",
    "I like AI",
    "AI is amazing"
]
```

```python
#Create an empty set
vocabulary = set()
```

```python
# Read each sentence in the corpus
for sentence in corpus:
    print(sentence)
```

```python
# Split each sentence
for sentence in corpus:
    words = sentence.split()
    print(words)
```

```python
# Add the words to the set
vocabulary.update(words)
```

```python
# Read every sentence
for sentence in corpus:
    
    # Split sentence into words
    words = sentence.split()
    
    # Add words to vocabulary
    vocabulary.update(words)
```

```python
# Print vocabulary
print("Vocabulary")

for word in vocabulary:
    print(word)
```

```python
# Print vocabulary size
print("Vocabulary Size =", len(vocabulary))
```
