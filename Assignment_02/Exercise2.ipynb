```python
corpus = [
    "I like NLP",
    "I like AI",
    "AI is amazing"
]

# 1. Create a vocabulary using a set
vocabulary = set()

# 2. Read every sentence
for sentence in corpus:
    # 3. Split into words
    words = sentence.split()
    
    # Add the words to the set (duplicates are automatically ignored)
    vocabulary.update(words)

# 4. Print the vocabulary and its size
print("Vocabulary")
for word in sorted(vocabulary):
    print(word)

print(f"\nVocabulary Size = {len(vocabulary)}")
```
