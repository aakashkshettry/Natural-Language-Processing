# Exercise 1: Word Frequency Counter

**1. Objective**

The objective is to count how many times each word appears in a sentence.

```python
#store the sentence in a Python variable:
text = "I like NLP I like Python NLP is fun"
```

```python
# Split the sentence into words
words = text.split()

print(words)
```

```python
#Create an empty dictionary
word_count = {}
```

```python
#Count each word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
```

```python
#Print the dictionary
print("Word Frequency:")
print(word_count)
```
