## Introduction

Part-of-Speech (POS) tagging is an important Natural Language Processing
technique in which each word in a sentence is assigned a grammatical
category such as noun, verb, adjective, adverb, pronoun, or preposition.

In this assignment, an English news article is analyzed using POS tagging.
The main focus is on extracting nouns and verbs from the article.

The Natural Language Toolkit (NLTK) is used for tokenization and
Part-of-Speech tagging, while Pandas is used to organize and export the
results.

**Importing Libraries**

```python
import nltk
import pandas as pd
import re
import matplotlib.pyplot as plt
```

**Download NLTK Resources**

These resources are required for word tokenization and POS tagging.

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

**Load the News Article**

This gives us a basic idea about the size of the article.

```python
# Load the news article from the text file

with open("news_article.txt", "r", encoding="utf-8") as file:
    text = file.read()

print("Article loaded successfully.")
print("\nArticle:")
print(text)
```

**Inspect the Article**

The regular expression replaces multiple spaces, tabs, and line breaks with a single space.

```python
print("Number of characters:", len(text))
print("Number of words before tokenization:", len(text.split()))
```

**Text Preprocessing**

```python
# Remove unnecessary whitespace

text = re.sub(r'\s+', ' ', text).strip()

print(text)
```

**Tokenization**

The article is converted from a continuous sequence of text into individual tokens.

```python
# Tokenize the article into individual words/tokens

tokens = nltk.word_tokenize(text)

print("Total number of tokens:", len(tokens))
print("\nFirst 30 tokens:")
print(tokens[:30])
```

**POS Tagging**

```python
# Perform Part-of-Speech tagging

pos_tags = nltk.pos_tag(tokens)

print("First 30 POS-tagged tokens:\n")

for word, tag in pos_tags[:30]:
    print(f"{word:20} {tag}")
```

**Identify Noun and Verb Tags**

```python
# POS tags representing nouns

noun_tags = [
    'NN',    # singular noun
    'NNS',   # plural noun
    'NNP',   # proper singular noun
    'NNPS'   # proper plural noun
]

# POS tags representing verbs

verb_tags = [
    'VB',    # base form
    'VBD',   # past tense
    'VBG',   # gerund/present participle
    'VBN',   # past participle
    'VBP',   # non-3rd person singular present
    'VBZ'    # 3rd person singular present
]
```

**Extract Nouns and Verbs**

```python
# Extract nouns and verbs

noun_verb_data = []

for word, tag in pos_tags:
    if tag in noun_tags or tag in verb_tags:
        noun_verb_data.append((word, tag))

print("Total nouns and verbs extracted:", len(noun_verb_data))

print("\nFirst 30 extracted words:")
print(noun_verb_data[:30])
```

**Create a DataFrame**

```python
df = pd.DataFrame(
    noun_verb_data,
    columns=["Word", "POS_Tag"]
)

df.head(20)
```

**Add Noun/Verb Category**

```python
def get_pos_category(tag):

    if tag in noun_tags:
        return "Noun"

    elif tag in verb_tags:
        return "Verb"

    return "Other"


df["POS_Category"] = df["POS_Tag"].apply(get_pos_category)

df.head(20)
```

**Display All Extracted Nouns**

```python
nouns = df[df["POS_Category"] == "Noun"]

print("Extracted Nouns:")
display(nouns)
```

**Display All Extracted Verbs**

```python
verbs = df[df["POS_Category"] == "Verb"]

print("Extracted Verbs:")
display(verbs)
```

**Count Nouns and Verbs**

```python
noun_count = len(nouns)
verb_count = len(verbs)

print("Total Nouns:", noun_count)
print("Total Verbs:", verb_count)
```

**POS Frequency Analysis**

```python
print("Frequency of POS Categories:")
print(df["POS_Category"].value_counts())
```

```python
print("\nFrequency of Individual POS Tags:")
print(df["POS_Tag"].value_counts())
```

**Save Results to CSV**

```python
output_file = "AakashBhusal_POS_01.csv"

df.to_csv(output_file, index=False)

print(f"CSV file successfully saved as: {output_file}")
```
