## Introduction

Named Entity Recognition (NER) is an important Natural Language Processing
technique used to identify specific entities mentioned in a text and
classify them into predefined categories.

Common named entity categories include:

- PERSON – names of people
- ORG – organizations
- GPE – geopolitical entities such as countries, cities, and states
- DATE – dates or periods of time
- MONEY – monetary values
- LOC – locations
- EVENT – named events

In this assignment, NER is performed on a real-world English news article
using the spaCy Natural Language Processing library.

The recognized entities and their corresponding entity types are extracted
and saved into a CSV file for further analysis.

**Import Libraries**

```python
import spacy
import pandas as pd
import matplotlib.pyplot as plt
```

**Load the spaCy Model**

```python
# Load the pretrained English language model

nlp = spacy.load("en_core_web_sm")

print("spaCy English model loaded successfully.")
```

**Load the News Article**

```python
# Load the news article

with open("news_article.txt", "r", encoding="utf-8") as file:
    text = file.read()

print("Article loaded successfully.")
print("\nArticle:")
print(text)
```

**Inspect the Article**

```python
print("Number of characters:", len(text))
print("Number of words:", len(text.split()))
```

**Apply Named Entity Recognition**

```python
# Process the article using spaCy

doc = nlp(text)

print("Named Entity Recognition completed.")
```

**Display Recognized Entities**

```python
print("Recognized Named Entities:\n")

for ent in doc.ents:
    print(f"{ent.text:30} {ent.label_}")
```

**Extract Entity and Entity Type**

```python
# Extract entities and their labels

entities = []

for ent in doc.ents:
    entities.append((ent.text, ent.label_))

print("Total entities extracted:", len(entities))
```

**Create DataFrame**

```python
df = pd.DataFrame(
    entities,
    columns=["Entity", "Entity_Type"]
)

display(df)
```

**Remove Duplicate Entities**

```python
unique_entities = df.drop_duplicates()

print("Total entity mentions:", len(df))
print("Unique entity mentions:", len(unique_entities))

display(unique_entities)
```

**Count Entity Types**

```python
entity_type_counts = df["Entity_Type"].value_counts()

print("Entity Type Counts:")
print(entity_type_counts)
```

**Display Entity Counts as a DataFrame**

```python
summary_df = (
    df["Entity_Type"]
    .value_counts()
    .reset_index()
)

summary_df.columns = ["Entity_Type", "Count"]

display(summary_df)
```

**Find Entities by Category**

```python
persons = df[df["Entity_Type"] == "PERSON"]

print("PERSON entities:")
display(persons)
```

```python
organizations = df[df["Entity_Type"] == "ORG"]

print("ORG entities:")
display(organizations)
```

```python
gpe = df[df["Entity_Type"] == "GPE"]

print("GPE entities:")
display(gpe)
```

```python
dates = df[df["Entity_Type"] == "DATE"]

print("DATE entities:")
display(dates)
```

**Save Entities to CSV**

```python
output_file = "AakashBhusal_NER_01.csv"

df.to_csv(output_file, index=False)

print(f"NER results successfully saved as: {output_file}")
```
